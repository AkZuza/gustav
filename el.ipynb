```python
# Connect to neo4j database instance
from neo4j import GraphDatabase
URI = "neo4j://localhost"
AUTH = ("neo4j", "byteme@2023")
def connect_database():
    with GraphDatabase.driver(URI, auth=AUTH) as ndriver:
        global driver
        driver = ndriver
        driver.verify_authentication()
        driver.verify_connectivity()
        print("Connected to the database")

connect_database()
```

```python
# Creates a transaction
def create_transaction(payer_id: str, recipient_id: str, amount: float):
    global driver
    driver.execute_query(f'MATCH (p:Account), (r:Account) WHERE p.acc={payer_id} AND r.acc={recipient_id} CREATE (p)-[t:PAYED {{amount: {amount }}}]->(r)')
    pass
```

```python
# Generates random transactions between any two people
# Do not put a large number, that would be stupid
# We need more people and accounts


ids = list()
records, s, k = driver.execute_query("MATCH (a:Account) RETURN a.acc as acc")
assert records is not None
for account in records:
    print(account["acc"])
    ids.append(account["acc"])

for i in range(50):
    payer_id = random.choice(ids)
    recp_id = random.choice(ids)
    while(payer_id == recp_id):
        recp_id = random.choice(ids)

    amount = random.random()
    create_transaction(payer_id, recp_id, amount)
```
